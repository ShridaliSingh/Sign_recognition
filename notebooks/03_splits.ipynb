```python
import json
import numpy as np
import pandas as pd

df = pd.read_csv("../data/kaggle/train.csv")
signers = np.sort(df["participant_id"].unique())
rng = np.random.default_rng(42)
shuffled = rng.permutation(signers)

test = sorted(shuffled[:4].tolist())
val = sorted(shuffled[4:7].tolist())
train = sorted(shuffled[7:].tolist())

for name, ids in [("train", train), ("val", val), ("test", test)]:
    part = df[df["participant_id"].isin(ids)]
    print(name, len(ids), "signers |", len(part), "sequences |", part["sign"].nunique(), "signs")

with open("../training/splits.json", "w") as f:
    json.dump({"train": train, "val": val, "test": test}, f, indent=2)
```
